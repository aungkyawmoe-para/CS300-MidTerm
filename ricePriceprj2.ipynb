```python
processed_riceDF['Price_per_kg']= pd.to_numeric(processed_riceDF['Price_per_kg'],errors='coerce')
processed_riceDF['Date']= pd.to_datetime(processed_riceDF['Date'],format='%d-%m-%Y')
processed_riceDF.dropna(subset='Date',inplace=True,ignore_index=True)
processed_riceDF= processed_riceDF.sort_values(by='Date')
processed_riceDF['Rainfall_mm']= processed_riceDF['Rainfall_mm'].ffill().bfill()
processed_riceDF['Temperature_C']= processed_riceDF['Temperature_C'].ffill().bfill()
```

```python
processed_riceDF['Date_Month']= processed_riceDF['Date'].dt.month.astype(str)
```

```python
processed_riceDF.drop(columns=['Date','Date_Month'],inplace=True)
```

```python
cat_cols = processed_riceDF.select_dtypes('str').columns
num_cols = processed_riceDF.select_dtypes('number').columns
```

```python
for col in cat_cols:
    print(processed_riceDF[col].unique())
```

```python
def filledUnknown_and_mean(df):
    for col in cat_cols:
        df[col]=df[col].fillna('Unknown')
    for col in num_cols:
        df[col]= df[col].fillna(df[col].mean())
    return df

def dropall(df):
    return df.dropna(ignore_index=True)

def drop_naCats_useMean(df):
    df = df.dropna(subset=cat_cols)
    for col in num_cols:
        df[col]= df[col].fillna(df[col].mean())
    return df

def filledmode_and_mean(df):
    for col in cat_cols:
        df[col]=df[col].fillna(df[col].mode()[0])
    for col in num_cols:
        df[col]= df[col].fillna(df[col].mean())
    return df

def filledmode_and_drop(df):
    for col in cat_cols:
        df[col]=df[col].fillna(df[col].mode()[0])
    df = df.dropna(subset=num_cols,ignore_index=True)
    return df

def filledUnknown_and_drop(df):
    for col in cat_cols:
        df[col]= df[col].fillna('Unknown')
    df = df.dropna(subset=num_cols,ignore_index=True)
    return df
```

```python
def standardizeCats(df):
    for col in cat_cols:
        df[col]= df[col].str.strip().str.title()
    return df
```

```python
processed_riceDF = standardizeCats(processed_riceDF)
processed_riceDF_cleanLab = processed_riceDF.dropna(subset='Price_per_kg',ignore_index=True)
```

```python
def outlier_replacer(df):
    for col in ['Demand_Tons','Rainfall_mm','Supply_Tons','Temperature_C']:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR  # Using 1.5*IQR for outliers
        upper_bound = Q3 + 1.5 * IQR
        
        #Fill here
        outliers = df.loc[(df[col]<lower_bound)|(df[col]>upper_bound)].shape[0]

        
        print(f"  {col}: {outliers} outliers detected")
        print(f"    Range before: [{df[col].min():.2f}, {df[col].max():.2f}]")
        
        # Fill here
        df[col]= df[col].clip(lower=lower_bound,upper=upper_bound)
        print(f"    Range After: [{df[col].min():.2f}, {df[col].max():.2f}]")
    return df
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsRegressor
```

```python
df_filledoutlier = pRDF_MM2
```

```python
X = df_filledoutlier.drop(columns='Price_per_kg')
y = df_filledoutlier['Price_per_kg']
X_cat_cols = X.select_dtypes('str').columns
X_num_cols = X.select_dtypes('number').columns
```

```python
df_filledoutlier.corr(numeric_only=True)
```

```python
onehot = OneHotEncoder(sparse_output=False,handle_unknown='ignore')
std = StandardScaler()
lrg = LinearRegression()

preprocess_track = ColumnTransformer(
    [('encoding',onehot,X_cat_cols),
     ('scaling',std,X_num_cols)
     ]
)

pline = Pipeline([
    ('Preprocessing',preprocess_track),
    ('Dimension Reduction',PCA(n_components=3)),
    ('Linear Regression',lrg)

])

pline2 = Pipeline([
    ('Preprocessing',preprocess_track),
    ('Linear Regression',lrg)

])
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,shuffle=True,random_state=42)
```

```python
X_train.shape
```

```python
pline.fit(X_train,y_train)
y_test_hat = pline.predict(X_test)
pline2.fit(X_train,y_train)

y_test_hat2 = pline2.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error
import numpy as np
```

```python
pline_test_mae=mean_absolute_error(y_test,y_test_hat)
pline_test_mse=mean_squared_error(y_test,y_test_hat)
pline_test_rmse = root_mean_squared_error(y_test,y_test_hat)
print(f"Mean absolute error {np.round(pline_test_mae,2)}")
print(f"Mean square error {np.round(pline_test_mse,2)}")
print(f'Root Mean Squared Error {np.round(pline_test_rmse,2)}')
```

```python
pline2_test_mae=mean_absolute_error(y_test,y_test_hat2)
pline2_test_mse=mean_squared_error(y_test,y_test_hat2)
pline2_test_rmse = root_mean_squared_error(y_test,y_test_hat2)
print(f"Mean absolute error {np.round(pline2_test_mae,9)}")
print(f"Mean square error {np.round(pline2_test_mse,9)}")
print(f'Root Mean Squared Error {np.round(pline2_test_rmse,2)}')
```

```python
print(y_test.describe())
```

```python
kNRe = KNeighborsRegressor(n_neighbors=5)
pline3 = Pipeline([
    ('Preprocessing',preprocess_track),
    ('Dimension Reduction',PCA(n_components=3)),
    ('KNN Regression',kNRe)

])

pline4 = Pipeline([
    ('Preprocessing',preprocess_track),
    ('KNN Regression',kNRe)

])
```

```python
pline3.fit(X_train,y_train)
y_test_hat3 = pline3.predict(X_test)
pline4.fit(X_train,y_train)

y_test_hat4 = pline4.predict(X_test)
```

```python
pline3_test_mae=mean_absolute_error(y_test,y_test_hat3)
pline3_test_mse=mean_squared_error(y_test,y_test_hat3)
pline3_test_rmse = root_mean_squared_error(y_test,y_test_hat3)
print(f"Mean absolute error {np.round(pline3_test_mae,2)}")
print(f"Mean square error {np.round(pline3_test_mse,2)}")
print(f'Root Mean Squared Error {np.round(pline3_test_rmse,2)}')
```

```python
pline4_test_mae=mean_absolute_error(y_test,y_test_hat4)
pline4_test_mse=mean_squared_error(y_test,y_test_hat4)
pline4_test_rmse = root_mean_squared_error(y_test,y_test_hat4)
print(f"Mean absolute error {np.round(pline4_test_mae,2)}")
print(f"Mean square error {np.round(pline4_test_mse,2)}")
print(f'Root Mean Squared Error {np.round(pline4_test_rmse,2)}')
```

```python
from sklearn.metrics import r2_score

print(r2_score(y_test,y_test_hat))
print(r2_score(y_test,y_test_hat2))
print(r2_score(y_test,y_test_hat3))
print(r2_score(y_test,y_test_hat4))
```
